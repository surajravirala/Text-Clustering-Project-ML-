```python
import pandas as pd
#PART (1) ==> LOADING DATA
#Load the Excel file into a DataFrame named 'name'

name= pd.read_excel('Clustering Project (1).xlsx')
#'display' usually returns None, so Doc_file will be None
Doc_file=display(name.head())
print(Doc_file)
```

Test Cleaning

```python
#PART (2) ==> TEXT CLEANING
import nltk
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
#Initialize the Lemmatizer to find the root/lemma of words
wordnet=WordNetLemmatizer()

corpus=[]
for i in range(len(name)):
  # Step 1: Remove all characters except letters (a-z, A-Z) using Regular Expressions
  rp=re.sub('[^a-zA-z]'," ",name["Article"][i])
  # Step 2: Convert text to lowercase for uniformity
  rp=rp.lower()
  # Step 3: Tokenize (split into a list of words)
  rp=rp.split()
  # Step 4: Remove stopwords (like 'the', 'is') and Lemmatize (e.g., 'running' -> 'run')
  rp=[wordnet.lemmatize(word) for word in rp if not word in (stopwords.words("english"))]
  # Step 5: Join words back into a single string and add to our corpus list
  rp=" ".join(rp)

  corpus.append(rp)

print(corpus)
```

```python
# PART (3) ==> VECTORIZATION
from sklearn.feature_extraction.text import CountVectorizer
#Initialize CountVectorizer to convert text into a matrix of token counts (Bag of Words)
cv_excel=CountVectorizer()
#Fit the model and transform our text corpus into a numerical array
x=(cv_excel.fit_transform(corpus).toarray())

x
```

```python
x.shape
```

```python
# PART (4) ==> LDA (Latent Dirichlet Allocation)
from sklearn.decomposition import LatentDirichletAllocation
# Initialize LDA to find 10 distinct topics/groups
model=LatentDirichletAllocation(n_components=10)
# Train the model on our numerical data 'x'
model.fit(x)
```

```python
the_result=model.transform(x)# Transform the data to get the probability of each document belonging to each of the 10 topics
```

```python
the_result[0]
```

```python
the_result[0].argmax()
```

Combining with Orginal Data

```python
# Use argmax(axis=1) to find the index of the highest probability (the dominant group) for each row
name['group']=the_result.argmax(axis=1)
#Final display of the DataFrame with the new 'group' column
name
```

```python
name[name['group']==9]
```

Scaling

* We should Apply Scaling When The Column's Does not Same Size off Numeric Values
* Bring All Columns to Same Level Magnitude Because Machine think Big Number Column is Not Important
1. Min-Max Scaling
2. Standard Scaling




```python
# Min-Max Scaling
import pandas as pd

df=pd.DataFrame({"x":[10,20,30,40,50]})
df
```

```python
from sklearn.preprocessing import MinMaxScaler
m_and_m=MinMaxScaler()

df["x_mm"]=m_and_m.fit_transform(df[["x"]])  # where The values are Calculated Based on Formula X-Xmin/Xmax-Xmin
df
```

```python
# Standard Scaling
from sklearn.preprocessing import StandardScaler
s_and_s=StandardScaler()

df["x_ss"]=s_and_s.fit_transform(df[["x"]])   # where The values are Calculated Based on Formula X-mean/Standard_dev
df
```
